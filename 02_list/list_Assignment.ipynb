# Assignment 6: List Methods

Create a list of random numbers and sort it in ascending and descending order. Remove the duplicates from the list and print the modified list.


```python
import random
random_numbers = [random.randint(1, 100) for _ in range(16)]
print("Random numbers are:- ", random_numbers)
ascending_num = sorted(random_numbers)
print("Sorted random numbers are:- ", ascending_num)

descending_num = sorted(random_numbers, reverse=True)
print("Sorted random numbers in descending order are:- ", descending_num)

duplicate_num = list(set(random_numbers))
print("Random numbers after removing duplicates are:- ", duplicate_num)
```

# Assignment 8: List of Dictionaries

Create a list of dictionaries where each dictionary represents a student with keys 'name' and 'score'. Sort the list of dictionaries by the 'score' in descending order and print the sorted list.

```python
dict = [{"name": "Alice", "age": 30, "score": 85}, {"name": "Bob", "age": 25, "score": 90}, {"name": "Charlie", "age": 35, "score": 75}]

sorted_students = sorted(dict, key=lambda x: x['score'], reverse=True)
print("Students sorted by score in descending order are:- ", sorted_students)
for student in sorted_students:
    print(student['name'], student['score'])
```

# Assignment 9: Matrix Transposition

Write a function that takes a 3x3 matrix (nested list) as input and returns its transpose. Print the original and transposed matrices.

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print("original matrix is:- ")
for i in matrix:
    print(i)

def transpose(matrix):
    transposed = []
    for i in range(len(matrix[0])):
        row = []
        for j in range(len(matrix)):
            row.append(matrix[j][i])
        transposed.append(row)
    return transposed

transposed_matrix = transpose(matrix)
print("Transposed matrix is:- ")
for i in transposed_matrix:
    print(i)
```

# Assignment 10: Flattening a Nested List

Write a function that takes a nested list and flattens it into a single list. Print the original and flattened lists.

```python
nested_list = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flattened_list = [item for sublist in nested_list for item in sublist]
print("Flattened list is:- ", flattened_list)
```

# Assignment 14: List Rotation

Write a function that rotates a list by n positions. Print the original and rotated lists.

```python
def rotate(list, n):
    return list[n:] + list[:n]

original_list = [1, 2, 3, 4, 5]
rotated_list = rotate(original_list, 2)
print("Original list is:- ", original_list)
print("Rotated list by 2 positions is:- ", rotated_list)
```

# Assignment 15: List Intersection

Write a function that takes two lists and returns a new list containing only the elements that are present in both lists. Print the intersected list.

```python
lst1 = [1, 2, 3, 5 , 7, 6]
lst2 = [1, 2, 6, 9 , 12 , 5]

def common_elements(lst1, lst2):
    return [x for x in lst1 if x in lst2]

print("Common elements between lst1 and lst2 are:- ", common_elements(lst1, lst2))
```
